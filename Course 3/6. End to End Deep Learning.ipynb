# End - to - End DL

> 📜📜
>
> ***End-to-end deep learning is a method where a single neural network is trained to learn all steps of a complex task by mapping <u>raw input</u> directly to desired output, eliminating the need for handcrafted intermediate processing or features.***

<img src="./images/e2e-dl-ex.png">

**When it is good**:
- When the data for the task is huge. And it sounds **obvious** to use that just looking at the feature, we can get the output.

**When it's not good**:
- Don't blindly apply the E2E on any task thinking that **it's smart**. The DL models are not magic. 
- We need hella data.
- In the **bone example** we are trying to find the age of the child based on the x-ray.
- Here, it makes more sense to first segment the bones, calculate some length and then finally age. So you know -- E2E may not make sense here "just by looking at the x-ray".
