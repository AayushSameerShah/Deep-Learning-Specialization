## Satisficing and Optimizing Metrics

- Sometimes it’s hard to combine all evaluation goals into a **single number metric** (like accuracy).
- Instead of combining disparate metrics (e.g., accuracy and running time) into a weighted sum, you can use:
  - **One optimizing metric:** The metric you want to maximize or minimize as much as possible (e.g., accuracy).
  - **One or more satisficing metrics:** Metrics that just need to **meet a threshold** (e.g., running time ≤ 100 ms). Beyond that, better values don’t matter much.

***

### Example: Cat Classifier

- Three classifiers with accuracies and running times:

| Classifier | Accuracy | Running Time |
|------------|-----------|---------------|
| A          | 90%       | 80 ms         |
| B          | 92%       | 95 ms         |
| C          | 95%       | 1500 ms       |

- Running time is the **satisficing metric** (must be ≤ 100 ms).
- Accuracy is the **optimizing metric** (want to maximize).
- The best choice is **Classifier B**: highest accuracy with acceptable running time.

***

### Example: Wake Word Detection

- Accuracy: Probability of correctly waking up for trigger words (Optimizing metric).
- False positives: Max 1 per 24 hours (Satisficing metric).

***

### Generalizing to $$N$$ Metrics

- Pick **1 optimizing metric.**
- Pick **$$N-1$$ satisficing metrics.**
- Algorithms must first satisfy all satisficing metrics.
- Among those, pick the best optimizing metric value.

***

### Practical Insights

- This method prevents artificial or arbitrary combining metrics via weighted sums.
- Improves decision making by setting clear expectations and priorities.
- Helps speed up model selection and tuning.

***

### Reminder

- Always evaluate these metrics on training, dev, or test sets.
- Adjust dev/test sets and evaluation metrics if they do not reflect your real-world goals or business needs.

***

**Summary:**  
Using optimizing and satisficing metrics helps balance multiple performance goals sensibly, allowing you to maximize what matters most while ensuring other critical constraints are met.
