# Multi task.

- This is no longer a **single, simple, logistic regression** in which you have single label for each example.
- Here, you have multiple labels for each examples.

<img src="./images/m-task.png">

### The loss function

It is easy AF. It just adds another term which is "sum over all labels".

$$ \frac{1}{m} \sum_{i=1}^m \sum_{j=1}^4 \mathcal{L}(\hat{y}_j^{(i)}, y_j^{(i)}) $$

## Benefits

- Say if you had 100 tasks, and each only had 1,000 labeled examples, traditional approaches would train a **separate neural network for each task**, resulting in **limited data per network**.
- With multitask learning, you ***share information between tasks***, allowing the network to effectively learn from all 100,000 examples across tasks—improving generalization for each individual task.
- **Each task benefits** from data and representations learned from the other tasks, which acts like a regularizer and can help the model do better than training separately.
- Multitask learning is **particularly effective when every task has a similar** (not vastly different) amount of data, and the tasks share low-level features (for example, pedestrian, car, stop sign, and traffic light detection in autonomous driving images all rely on similar underlying vision features).

---

- In practice, multitask learning is often less common than transfer learning—but can be very valuable when you have many related tasks where labeling everything for every task in every sample is impractical.
- The approach also allows you to leverage partially labeled datasets efficiently, by ignoring missing labels during loss computation.
