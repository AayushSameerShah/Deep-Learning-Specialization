- In deep learning we can:
    - Independently reduce bias
    - Independently reduce variance

### 🔗 For each type of problem we have seperate solutions:


#### **Chain of Assumptions in Machine Learning**

For a model to succeed in the real world, it must pass through several stages—each with its own challenges and solutions:


#### **`1.` Fit Training Set Well on Cost Function**
- **Goal:** Model achieves low error on the training data (*ideally close to human-level performance*).
- **If not:** The model has high bias (underfitting).
- **Solutions:**
  - **Make the network bigger** (increase capacity)
  - **Switch to a better optimizer** (e.g., Adam)
  - **Other: Tune architecture, train longer, etc.**

***

#### **`2.` Fit Dev (Validation) Set Well on Cost Function**
- **Goal:** Model generalizes from training to validation/dev set.
- **If not:** The model has high variance (overfitting).
- **Solutions:**
  - **Apply regularization** (L2, dropout, data augmentation)
  - **Increase the size of the training set** (more data)

***

#### **`3.` Fit Test Set Well on Cost Function**
- **Goal:** Model generalizes from validation/dev set to test set.
- **If not:** The dev and test sets may not come from the same distribution.
- **Solutions:**
  - **Increase the size of the dev/validation set** to ensure representativeness
  - **Make sure the dev and test sets are similar**

***

#### **`4.` Perform Well in the Real World**
- **Goal:** Model performs well on real-world data, e.g., for actual users.
- **If not:** There is a distribution shift between test data and real-world data (data mismatch).
- **Solutions:**
  - **Change the dev/test set or cost function** to better reflect real-world tasks and priorities

***

**Summary Table:**

| Stage                          | Main Problem             | Typical Solutions                             |
|---------------------------------|--------------------------|-----------------------------------------------|
| Fit training set                | High bias (underfit)     | Bigger network, better optimizer              |
| Fit dev set                     | High variance (overfit)  | Regularization, more training data            |
| Fit test set                    | Data mismatch (dev/test) | Bigger/representative dev set                 |
| Perform well in real world      | Data mismatch (deploy)   | Align dev/test set, adjust cost function      |

***

> *Different problems in the ML pipeline require different targeted solutions--so diagnose carefully and fix the right stage!*
