## The Power of the Single Number

- In machine learning, whether tuning hyperparameters, testing new algorithms, or building systems, having a **single real-number evaluation metric** greatly accelerates progress.
- This metric lets you quickly compare models or experiments and decide which one performs better.
- Example: For a cat classifier, instead of separately comparing **precision and recall** (which often trade off), use a combined metric like the **F1 score** (harmonic mean of precision and recall) for a clear winner.
- Having two numbers (like precision and recall) makes it hard to judge which model is better when one excels in one metric and another in the other.
- Besides classification metrics, when evaluating across multiple categories or geographies, track the per-category metric *and* a single aggregate metric (often just the average) to simplify model comparison.
- A **well-defined development (dev) set** plus a **single-number evaluation metric** makes iterating experiments faster and more efficient.
- The ML workflow is iterative: generate an idea, implement it, test using your unified metric, and keep improving.
- This approach helps teams quickly identify promising models instead of getting lost in multi-dimensional metrics.

***

**Key takeaway:**  
> Define and use a single-number metric for your task early on to speed up model iteration, comparison, and decision-making.
