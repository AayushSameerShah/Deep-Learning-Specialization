# We will be simple here.

1. **Fully Connected Networks (ANNs) have major limitations for images:**
    - They require an *enormous number of parameters* (weights) even for small images, which means:
      - The model becomes very large and slow to train.
      - It easily *overfits* to training data and fails to generalize well.
    - They do *not leverage the local spatial structure* of images.
      - Every pixel is connected to every output neuron, ignoring the fact that nearby pixels are more related.
      - This causes the network to lose important “local context” information like edges or textures.


2. **CNNs use convolutions to focus on local features:**
    - CNNs apply small `filters` that *slide over* the image to learn *local patterns* like edges, corners, and textures.
    - This helps the network first capture *low-level features* and then combine them into *higher-level features* (like shapes and objects) as you go deeper.
    - By restricting connections to small neighborhoods, CNNs explicitly *exploit the spatial structure* of images.
    

3. **Why convolution layers are more efficient and powerful:**
    - `Filters` are *reused* across the entire image (“weight sharing”), drastically reducing the number of parameters.
    - Because filters slide over the image, CNNs have *translation invariance*—they detect the same pattern anywhere in the image.
    - This results in smaller, faster, and more generalizable models, *better suited to visual tasks* than fully connected layers alone.
    
4. **Translation Variance:**
    - The picture of a cat is a picture of a cat! The **same filter** is used for all parts of the image, so the network becomes robust to the position of the cat.
    - This is the inherit benefit we get from the CNNs.

# 🖼️ Padding!

<img src="./images/padding.png">

## What is padding?

- When you do a convolution, the output becomes smaller than the input (e.g., 5×5 image with 3×3 filter → 3×3 output).  
- Padding **means adding extra pixels** (usually zeros) around the border of the image so that the filter can also “sit” on the edges properly.

You can imagine putting a picture on a slightly larger white paper by adding a white frame around it – that frame is the padding.


## Why do we need padding?

1. **To stop the image from shrinking too fast**  
   - Without padding, every conv layer shrinks height/width.  
   - After many layers, your feature map becomes tiny very quickly and you lose a lot of spatial information.  
   - Padding lets you keep the spatial size (in “same” padding) so you can build deep networks without collapsing the image.

2. To let edge pixels get a fair chance  
   - **Center pixels get used in many filter** positions, **but corner/edge pixels get used fewer** times.  
   - With padding, the filter can slide “outside” the original image, so border pixels are involved in just as many operations as central pixels.  
   - This means features near the edges are learned as well as those in the middle (important if the object is near the border of the image).
