# **How and where do these guys help?**:

## 🧠 What the fuck is a 1×1 convolution (in one line)

* **Pixel-wise fully connected layer across channels**
* Spatially kuch nahi dekhta, **sirf depth (channels) ke saath khelta hai**

---

## 🔥 WHEN to use 1×1 conv (use-cases)

### ✅ 1. **Channel Reduction (Dimensionality Reduction)**

* Jab feature map ka depth bahut zyada ho
  e.g. `56×56×256 → 56×56×64`
* **Why?**

  * Compute kam
  * Memory kam
  * Next heavy conv sasta ho jaata

🧠 *Think:* gym mein fat cut karna before photoshoot

---

### ✅ 2. **Channel Expansion**

* Jab network ko zyada expressive banana ho
  e.g. `28×28×64 → 28×28×256`
* **Why?**

  * Rich feature combinations
  * More representational power

🧠 *Think:* lean body pe muscles add karna

---

### ✅ 3. **Cross-Channel Interaction**

* Normal conv: spatial + channel mixing
* **1×1 conv: only channel mixing**
* Har pixel pe channels ka **learned weighted combo**

🧠 *Analogy:* same location, but brain decides
“edge + color + texture → THIS is an eye”

---

### ✅ 4. **Bottleneck Layers (VERY IMPORTANT)**

* Used in:

  * ResNet
  * Inception
  * MobileNet
* Pattern:

```
1×1 (reduce) → 3×3 (process) → 1×1 (expand)
```

**Why?**

* Heavy work cheap channels pe hota
* Accuracy same, compute massively reduced

🧠 *This is elite optimization shit.*

---

### ✅ 5. **Adding Non-Linearity WITHOUT changing spatial size**

* Each 1×1 conv followed by ReLU = extra depth
* More non-linear decision boundaries

🧠 *Depth > width sometimes*

---

### ✅ 6. **Replacing Fully Connected Layers (esp. in CNNs)**

* In classification heads:

```
Conv → 1×1 → Global Avg Pool
```

* Keeps spatial info longer
* Fewer parameters than FC

🧠 *Clean, elegant, no overfitting bullshit*

---

## 🚫 WHEN NOT to use 1×1 conv

* ❌ When spatial context matters
  (edges, corners, textures → need 3×3 or 5×5)
* ❌ As first layer on raw image
  (image needs spatial understanding first)

---

## ⚡ Why 1×1 conv is powerful (summary bullets)

* 🚀 Reduces computation **drastically**
* 🧩 Learns channel-wise feature combinations
* 🧠 Adds depth & non-linearity cheaply
* 🏗 Enables deep networks without blowing memory
* 🔑 Core building block of modern CNNs

---

## 📌 One brutal intuition (remember this)

> **3×3 looks around, <br>
> 1×1 thinks.**

## ✨ Also Knwon As: `"Network-in-Network"`
Why?

- Because while using 1x1 kernel, no size is changed, but the items are compressed or expanded.
- So **before moving to the next layer** the network "processes" the input and thinks about it in special way.
- This part is still a "mini network" -- as it does the kernel and activation thing.
