# Pooling and FN

The CNN is made of **ConvLayer**, **pooling** and **Fully Connected Layer**.

### Max Pooling | Don't skip read it.

<img src="./images/max-pooling.png">

**The pooling**:
- It gives the highely abstract features
- Reduces noise significantly
- Max pool only picks the max valeus per region.
- That means **only those features** which are **present** according to that kernel are coming in the pooling layer.
    - > ✨ According to the example above, the feature **doesn't exist** in the upper right corner, that's why it has lower value compared to the upper left corner.
