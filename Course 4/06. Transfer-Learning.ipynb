**The ones**:

* When we are working on the similar task -- we can use some pretrained network to save the computation and the effor to train from scratch.

* Generally we will have to run the training for **multiple epochs**.

* <u>So for the same input **we can save the tensors from the frozen layers** and then reuse them in other epochs.</u>
