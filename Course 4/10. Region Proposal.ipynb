# R-CNN
Regions with CNN

**Motivation**:
- We can use the CNN-aided-sliding window to detect the object.
- But it often leads with **empty windows** and may detect something evenif nothing is there.

<img src="./images/rcnn-1.png">

**Solution**:
- Select **a few windows** and only run the classifiers on those windows.
- Use the segmentation algorithm to do that.

<img src="./images/rcnn-2.png">

### 🥉 R-CNN (Region-based CNN)

**Core idea**:
- Detect objects by ***first proposing candidate regions***, then classifying each region using a CNN.

**Pipeline**:
- Region proposals via Selective Search (~2000 regions)
- Crop each region
- Resize to fixed size (e.g. 224×224)
- Run a **classic CNN** (with FC layers) on each region
- Classify + refine bounding box

**Good**:
- First CNN-based detector to significantly outperform traditional methods
- High detection accuracy (for its time)

**Bad**:
- Extremely slow (CNN run thousands of times per image)
- Not end-to-end trainable
- Heavy disk usage
- Uses **external region proposal algorithm**
- Does NOT share convolutional computation

### 🥈 Fast R-CNN

**Core idea**:
- Share convolutional computation across the whole image.

**Pipeline**:
- Run CNN **once** on the full image → feature map
- Use region proposals (still Selective Search)
- Apply RoI Pooling on feature map
- Single network predicts:
  - class probabilities
  - bounding box refinements

**Good**:
- Much faster than R-CNN
- End-to-end training
- CNN computation shared
- Better memory efficiency

**Bad**:
- Still bottlenecked by Selective Search
- Region proposals are not learned


### 🥇 Faster R-CNN

**Core idea**:
- Learn region proposals using a neural network.

**Pipeline**:
- Backbone CNN produces feature map
- Region Proposal Network (RPN):
  - Slides over feature map
  - Uses anchor boxes
  - Predicts objectness + proposal boxes
- RoI Pooling / RoI Align
- Detection head:
  - Class prediction
  - Bounding box regression

**Good**:
- Fully end-to-end
- No external region proposal algorithm
- High localization accuracy
- Excellent performance on small and overlapping objects
- Still a strong baseline in research & industry

**Bad**:
- Slower than one-stage detectors (YOLO, SSD)
- More complex architecture


### Key Differences (Summary)

- R-CNN:
  - CNN runs per region → very slow

- Fast R-CNN:
  - CNN runs once → regions pooled from feature map

- Faster R-CNN:
  - Region proposals are learned (RPN)
  
---

### Comparison with YOLO

- R-CNN family:
  - Two-stage (propose → classify)
  - High accuracy
  - Slower inference

- YOLO:
  - One-stage (predict boxes directly)
  - Extremely fast
  - Slightly less precise localization (historically)
