**Lemme explain**:
- In the sliding window, normally -- you will have multiple levels of **granularity**.
- Model-1 detects the images with smaller frame (say expects 7x7 images only)
- Model-2 detects images with slightly larger frame (say expects 24x24 images)

> The bit issue is, for each single image, each model has to iterate multiple times (here for example 27 runs are required)

#### 🤔 But why are we doing this?
...because we want to "pinpoint" where is the required object in the image. For that we need to check each pixel!

## 🚗 Sliding Window — the old, dumb-but-honest way

Let’s say:

* You have a **car classifier**
* Input size = `64×64`

### What classic sliding window does:

1. Crop a `64×64` patch from the image
2. Run the **entire CNN**
3. Move window by stride (say 8 pixels)
4. Repeat **thousands of times**
5. Repeat again for **bigger windows** (128×128, 256×256…)

### Problems:

* 💀 Same convolutions repeated again and again
* 💀 Separate models or resizing hacks
* 💀 Slow as fuck

This works… but it’s **brute force caveman shit**.
