### `1.3` VGG-16

<img src="./images/vgg-16.jpg">

**Acrhitecture**:
```
Image 
    -> CONV x2 -> POOL 
        -> CONV x2 -> POOL 
            -> CONV x3 -> POOL
                -> CONV x3 -> POOL
                    -> CONV x3 -> POOL
                        -> FC -> FC -> Output
```

**Why**:
- **16 layers with weights**
- Has 138M params
- > ***Halves the size of image each layer while doubles up the number of channels.***

# `2.` ResNet

<img src="./images/skip-connection.png">

**Why**?:
- It allows to **train deep, very deep nets**.


## Why do they work well? Like why?

<img src="./images/why-res-works.png">

**Why**:
- If you had a simple model say `10 layers`, then the output after 10 layers will be good okay?
- If you had complex model say with `13 layers`, then generally your weights will have vanishing stuff going on.
- So if you apple the residual block, you will **still have the same output** as you did in the 10 layers in the model with 13 layers!! Wowo!!

**Some things**:
1. Using residual connections **never hurts** performance.
2. In many cases it will **help**.
