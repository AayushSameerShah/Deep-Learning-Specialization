## **Softmax for Multi-class Classification**

### **Overview**

- **Softmax regression** extends logistic regression to handle more than two classes (`C` possible labels, e.g. cat, dog, chick, other).
- The output is a vector of probabilities for each class, always summing up to 1.

***

### **How Softmax Works**

1. **Output Layer Setup:**
   - For `C` classes, output layer has `C` units.
   - Each unit predicts probability of the input belonging to a specific class.

2. **Linear Combination:**
   $$
   z^{[L]} = W^{[L]} a^{[L-1]} + b^{[L]}
   $$
   - Here, $$ z^{[L]} $$ is the vector of scores (one score for each class).

3. **Softmax Activation:**
   - Apply exponentiation to each element of $$ z^{[L]} $$:  
     $$
     t_i = e^{z_i^{[L]}}
     $$
   - Normalize to turn scores into probabilities:
     $$
     a_i^{[L]} = \frac{t_i}{\sum_{j=1}^{C} t_j}
     $$
   - The output $$ \vec{a}^{[L]} $$ is your probability vector (`y_hat`).

4. **Example Calculation:**
   - Suppose $$ z^{[L]} = [5,\, 2,\, -1,\, 3] $$
   - $$ t = [e^5,\, e^2,\, e^{-1},\, e^3] \approx [148.4,\, 7.4,\, 0.4,\, 20.1] $$
   - Sum = 176.3
   - Final probabilities:
     - Class 0: $$ 148.4/176.3 = 0.842 $$
     - Class 1: $$ 7.4/176.3 = 0.042 $$
     - Class 2: $$ 0.4/176.3 = 0.002 $$
     - Class 3: $$ 20.1/176.3 = 0.114 $$

***

### **Intuition**

- **Softmax turns raw, unbounded scores (logits) into probabilities.**
- Every output is non-negative, and all outputs sum to 1.
- The highest score is likely (but not always) the predicted class.
- If two classes have similar scores, their probabilities will be close together.

***

### **Decision Boundaries** 🚨

- With no hidden layers, Softmax regression produces **linear decision boundaries** between classes (generalizing the idea from logistic regression).
- With hidden layers, the boundaries become non-linear and much more powerful.

***

### **Key Points**

- Use Softmax when predicting a **single label out of multiple classes**.
- Output is a **vector of class probabilities**:  
  $$ \mathbf{y_{\text{hat}}} = [p(\text{class 0}),\ p(\text{class 1}), \ldots, p(\text{class C - 1})] $$
- **Softmax layer**: Exp, sum, divide—gives normalized probabilities.
- **Each class's output** can be interpreted as:  
  "Probability that this input belongs to this class, given the input."

***

**In summary:**  
Softmax is crucial for multi-class classification, giving an easy way to predict and reason about probabilities for each possible class in a mutually exclusive problem.

> *With no hidden layers, Softmax regression produces **linear decision boundaries** between classes (generalizing the idea from logistic regression).*
