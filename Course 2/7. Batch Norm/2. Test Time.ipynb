# 🤔 We calculate and update the batch norm -- on each batch!?

### **How BatchNorm Handles Mean and Std During Training and Testing**

- During **training**, batch normalization computes the *mean* and *standard deviation* for each batch separately.
    - If you have `1000` examples and train with batch size `100`, each of the 10 batches will have **its own mean and std** for normalization.
    - This introduces *slight randomness* in how each batch is normalized, which (as noted before) adds some mild regularization.

- During **testing/inference**, you may not have a batch—sometimes only a single input example!
    - Calculating statistics just for one example makes no sense, and the normalization would be very unstable.

***

### **How do we handle this?**

- During training, batch norm **keeps track of a running (exponentially weighted -- EWMA) average** of the mean and std for each feature/channel.
    - Think of this as the *global* mean and std, slowly updated as the network sees more data.
- At test time, just use these **stored running statistics** to normalize new (possibly single) inputs.
    - This makes the network’s behavior at inference stable and independent of the specific batch it saw during training.

***

### **Summary Table**

| Phase          | Normalization stats         | Effect                                 |
|----------------|----------------------------|----------------------------------------|
| Training       | Batch mean & std (`per-batch`) | Slightly different for each batch; updates the running averages |
| Testing        | Running mean & running std (`global`) | Stable, batch-independent normalization |

***

### **Why is this important?**

- **Consistency:** Ensures that even if you feed in just *one* example at test time, the network normalizes it in a meaningful, stable way.
- **Performance:** Prevents weird results due to outlier or tiny test batches.

***

**Key point:**  
> *BatchNorm needs to “remember” how your data looked on average during training, so it can apply the same normalization at inference—no matter how little data it sees at test time!*
