# 🔥 Random DOESN'T mean UNIFORMLY RANDOM!

- For discrete structural choices like layers **uniform sampling** is appropriate because each option is a distinct architecture choice, not an order‑of‑magnitude scale ✅.
- For continuous scales like learning rate $\alpha \in [0.0001, 1.0]$, **uniform sampling over the linear interval** overemphasizes the high end (e.g., 0.1–1.0) and almost never tries tiny but crucial values like 0.0002 or 0.0005 ❌.  
- The fix: sample on a **log scale**, i.e., sample $u \sim \text{Uniform}[\log_{10}(a), \log_{10}(b)]$ then set $\alpha = 10^{u}$, which allocates equal probability mass per order of magnitude and explores small values effectively ✅.  

### Rule of thumb

- Use **uniform** for categorical/discrete sets and parameters without multi‑order variation.  
- Use **log‑uniform** for positive continuous scales spanning orders of magnitude (learning rate, weight decay, epsilon‑like stabilizers).  

### TL;DR

So, **Use Log Scale** for ranges like $[10^{-4}, 1]$ to fairly test small yet often optimal values, and **Uniform** for discrete architecture choices or narrow linear ranges.
